```python
import sympy as sp
```

## Limits

```python
x = sp.symbols('x')
f = 1/x
result = sp.limit(f, x, 'oo')
print(result)
```

## Euler's Number !

```python
n = sp.symbols('n')
f = (1 + 1/n) ** n
e = sp.limit(f, n, sp.oo)
print(e)
print(type(e))
print(e.evalf())
```

## Find Closed form for sum of 1st n numbers

```python
k, n = sp.symbols('k n')
s = sp.summation(k,(k, 1, n))
sf = s.factor()
print(s)
print(sf)



```

## Find Closed form for sum of 1st n squared numbers

```python
k, n = sp.symbols('k n')
s = sp.summation(k**2,(k, 1, n))
sf = s.factor()
print(s)
print(sf)
```

# Chain rule derivative

```python
x, y = sp.symbols('x y')
y_expr = x**2 + 1
z = y**3 - 2
dy_x = sp.diff(y_expr, x)
print(dy_x)
dz_y = sp.diff(z, y)
print(dz_y)
dz_dx = dz_y * dy_x
print(dz_dx.subs(y, y_expr))
```
