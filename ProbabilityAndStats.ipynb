```python
from scipy.stats import binom, beta
```

### binomial (k , n , p)

```python
n = 10
p = 0.9
for k in range (0, n+1):
    pkn = binom.pmf(k, n, p)
    print(f'#{k} {pkn}')
```

### Beta distribution for 8 successes and 2 failures what is the probability (p < 0.9)  cdf = cumulative density function



```python
d = beta.cdf(0.9, 8, 2)
print(f'probability of p being 90% or less: {d}')
print(f'probability of p being 90% or more: {1 - d}')
```

### Weighted mean

```python
samples = [90, 80, 63, 87]
weights = [.20, .20, .20, .40]
weighted_mean = sum(s*w for s, w in zip(samples, weights)) / sum(weights)
print(f'weighted mean: {weighted_mean} with {weights}')
# note the weights do not need to be percentages !
weights = [1, 1, 1, 4]
print(f'weighted mean: {weighted_mean} with {weights}')
```

### Median calculation : The advantage over mean is that any outlier will not affect the median, instead a big outlier will impact the mean.

```python

def median(sample):
    assert len(sample) > 0, "Sample cannot be empty"

    ss = sorted(sample)
    l = len(ss)
    mid_idx = l // 2
    mid_elem = (ss[mid_idx] + ss[mid_idx-1]) / 2 if l%2 == 0 else ss[mid_idx]
    return mid_elem

sample = [0, 1, 5, 7, 9, 10, 14]
print(f'median: {median(sample)} with {sample}')

sample = [0, 1, 5, 7, 7, 9, 10, 14]
print(f'median: {median(sample)} with {sample}')

sample = [0, 1, 5, 7, 9, 6, 10, 14]
print(f'median: {median(sample)} with {sample}')
```
